# Periapical Segmentation & Landmark Detection Training Pipeline

This notebook documents the training process for the two-stage deep learning pipeline used in the Bitewing Landmark Annotator.

## Overview

The pipeline consists of two distinct stages designed to handle the complexity of dental radiographs. The first stage, **Tooth Detection**, utilizes a YOLOv8 Oriented Bounding Box (OBB) model to identify and localize individual teeth within the full bitewing X-ray. Unlike standard bounding boxes, OBBs capture the rotation of each tooth, ensuring a precise fit that minimizes background noise and overlapping structures.

The second stage, **Landmark Localization**, focuses on identifying specific anatomical keypoints within each detected tooth. Using a UNet-based architecture, this stage processes the cropped and rectified regions of interest (ROIs) from Stage 1 to predict heatmaps for landmarks such as the Cemento-Enamel Junction (CEJ) and the alveolar bone crest.
